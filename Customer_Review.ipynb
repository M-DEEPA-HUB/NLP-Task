```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
stop_words = set(stopwords.words("english"))
df=pd.read_csv("IMDB_Dataset_CLEANED.csv")
print(df.isnull().sum())
print(df.duplicated().sum())
df["review"]=df["review"].str.lower()
pattern=r"<.*>"
df["review"]=df["review"].apply(lambda x: re.sub(pattern,"",x))
url=r"http:\s*://\S*"
df["review"]=df["review"].apply(lambda x: re.sub(url,"",x))
punc=r"[^\w\s]"
df["review"] = df["review"].apply(
    lambda x: re.sub(punc, "", x)
)
num=r"\d+"
df["review"] = df["review"].apply(lambda x: re.sub(num, "", x))
special_pattern = r"[^a-z\s]"

df["review"] = df["review"].apply(
    lambda x: re.sub(special_pattern, "", x)
)
df["tokenized_review"] = df["review"].apply(lambda x: x.split())
df["tokenized_review"] = df["tokenized_review"].apply(
    lambda words: [word for word in words if word not in stop_words]
)
df["cleaned_review"] = df["tokenized_review"].apply(
    lambda words: " ".join(words)
)
df["lemmatized_review"] = df["tokenized_review"].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)
df["cleaned_review"] = df["lemmatized_review"].apply(
    lambda words: " ".join(words) 
)
print(df[["review", "sentiment", "cleaned_review"]].head())
```
